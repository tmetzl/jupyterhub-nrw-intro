# Daten in Jupyter: Gapminder

In diesem Notebook liest du einen echten Datensatz ein, filterst ihn und zeichnest Diagramme. Du musst keinen Code schreiben. Du führst ihn aus, änderst einzelne Werte und schaust, was passiert.

**Der Datensatz:** [Gapminder](https://www.gapminder.org/) sammelt Kennzahlen zu allen Ländern der Welt. Wir nutzen einen bekannten Ausschnitt: 142 Länder, 1952 bis 2007 in Fünfjahresschritten, mit Lebenserwartung, Bevölkerung und Wirtschaftsleistung pro Kopf. Lizenz: CC BY 4.0.

## Einlesen

Die Bibliothek **pandas** ist das Standardwerkzeug für Tabellen in Python. `read_csv` liest die Datei, `head()` zeigt die ersten fünf Zeilen.

```python
import pandas as pd

df = pd.read_csv("daten/gapminder.csv")
df.head()
```

Die Spalten sind auf Englisch:

| Spalte | Bedeutung |
|---|---|
| `country` | Land |
| `year` | Jahr |
| `pop` | Bevölkerung |
| `continent` | Kontinent |
| `lifeExp` | Lebenserwartung in Jahren |
| `gdpPercap` | Bruttoinlandsprodukt pro Kopf in US-Dollar |

## Ein erster Überblick

`describe()` berechnet für jede Zahlenspalte Kennwerte wie Mittelwert, Minimum und Maximum.

**Aufgabe:** Lies aus der Ausgabe ab, wie viele Zeilen die Tabelle hat (Zeile `count`) und wie niedrig die niedrigste Lebenserwartung war (Zeile `min`, Spalte `lifeExp`).

```python
df.describe()
```

## Filtern und zeichnen

Jetzt holen wir ein einzelnes Land aus der Tabelle und zeichnen die Lebenserwartung über die Zeit.

Die Zeile `df[df["country"] == land]` liest sich als: "Gib mir alle Zeilen, in denen die Spalte `country` gleich `land` ist."

**Aufgabe:** Ändere `land` in ein anderes Land und führe die Zelle neu aus. Die Namen sind englisch, zum Beispiel `"France"`, `"Japan"`, `"Nigeria"` oder `"Brazil"`.

```python
land = "Germany"

auswahl = df[df["country"] == land]
auswahl.plot(
    x="year",
    y="lifeExp",
    marker="o",
    title=f"Lebenserwartung in {land}",
    xlabel="Jahr",
    ylabel="Lebenserwartung in Jahren",
    legend=False,
);
```

Falls du nicht weißt, wie ein Land geschrieben wird: Die nächste Zelle listet alle Länder auf.

```python
sorted(df["country"].unique())
```

## Alle Länder auf einen Blick

Das berühmte Gapminder-Diagramm: Jeder Punkt ist ein Land. Rechts ist reich, oben lebt man lang. Die Punktgröße zeigt die Bevölkerung, die Farbe den Kontinent.

Die Zelle definiert eine **Funktion** `zeige_jahr`, die das Diagramm für ein Jahr zeichnet, und ruft sie einmal für 2007 auf.

**Aufgabe:** Ändere das Jahr in der letzten Zeile, zum Beispiel auf `1952`, und führe die Zelle neu aus. Es gibt Daten für 1952, 1957, 1962, ... bis 2007.

```python
import matplotlib.pyplot as plt


def zeige_jahr(jahr):
    daten = df[df["year"] == jahr]

    fig, ax = plt.subplots(figsize=(9, 6))
    for kontinent, gruppe in daten.groupby("continent"):
        ax.scatter(
            gruppe["gdpPercap"],
            gruppe["lifeExp"],
            s=gruppe["pop"] / 1e6,
            alpha=0.6,
            label=kontinent,
        )

    ax.set_xscale("log")
    ax.set_xlim(200, 100_000)
    ax.set_ylim(20, 90)
    ax.set_xlabel("BIP pro Kopf in US-Dollar (logarithmisch)")
    ax.set_ylabel("Lebenserwartung in Jahren")
    ax.set_title(f"Gapminder {jahr}")
    ax.legend(title="Kontinent", loc="lower right")
    plt.show()


zeige_jahr(2007)
```

## Ein Slider

Statt das Jahr im Code zu ändern, bauen wir einen Schieberegler. `interact` aus **ipywidgets** nimmt unsere Funktion und erzeugt für den Parameter `jahr` automatisch einen Slider von 1952 bis 2007 in Fünferschritten.

**Aufgabe:** Zieh den Regler von links nach rechts und schau, wie die Länder nach oben rechts wandern.

```python
from ipywidgets import interact

interact(zeige_jahr, jahr=(1952, 2007, 5));
```
