Dieses Notebook läuft nicht mit Python, sondern mit **R**. Schau oben rechts: Dort steht der Name des Kernels. Derselbe Hub, dieselben Daten, eine andere Sprache.

Wir nutzen wieder die Gapminder-Daten und rechnen damit das, was man aus Statistikkursen kennt: Korrelation, t-Test, lineares Modell. Alles mit Bordmitteln von R, ohne Zusatzpakete.

## Einlesen

`read.csv` liest die Datei in einen **data.frame**, das Gegenstück zur pandas-Tabelle. Der Pfad beginnt mit `../`, weil dieses Notebook im Unterordner `zugaben/` liegt.

```r
df <- read.csv("../daten/gapminder.csv")
head(df)
```

## Überblick

`summary()` zeigt für jede Spalte Minimum, Quartile, Median, Mittelwert und Maximum.

```r
summary(df)
```

## Ein Jahr herausfiltern und zeichnen

`subset()` wählt Zeilen nach einer Bedingung aus. Die Tilde in `lifeExp ~ log(gdpPercap)` liest sich als "Lebenserwartung in Abhängigkeit vom logarithmierten BIP". Diese Formelschreibweise taucht in R überall auf, auch gleich beim Modell.

```r
daten_2007 <- subset(df, year == 2007)

plot(
  lifeExp ~ log(gdpPercap),
  data = daten_2007,
  xlab = "log(BIP pro Kopf)",
  ylab = "Lebenserwartung in Jahren",
  main = "Gapminder 2007",
  pch = 19,
  col = factor(continent)
)
legend("bottomright", legend = levels(factor(daten_2007$continent)), col = 1:5, pch = 19, title = "Kontinent")
```

## Korrelation

Wie stark hängen Wohlstand und Lebenserwartung zusammen? `cor.test()` liefert den Korrelationskoeffizienten, ein Konfidenzintervall und einen p-Wert.

**Aufgabe:** Ändere oben das Jahr in `daten_2007 <- subset(df, year == 2007)` auf `1952`, führe beide Zellen neu aus und vergleiche die Korrelation.

```r
cor.test(daten_2007$lifeExp, log(daten_2007$gdpPercap))
```

## t-Test

Unterscheidet sich die Lebenserwartung 2007 zwischen Europa und Amerika? Ein t-Test für zwei Gruppen, wieder in Formelschreibweise: `lifeExp ~ continent` heißt "Lebenserwartung, gruppiert nach Kontinent".

**Aufgabe:** Tausche `"Americas"` gegen `"Asia"` oder `"Africa"` und schau, wie sich der p-Wert verändert.

```r
zwei_kontinente <- subset(daten_2007, continent %in% c("Europe", "Americas"))

t.test(lifeExp ~ continent, data = zwei_kontinente)
```

## Lineares Modell

`lm()` schätzt eine Regressionsgerade, `summary()` zeigt die übliche Ausgabe mit Koeffizienten, Standardfehlern, p-Werten und R².

**Aufgabe:** Lies aus der Tabelle `Coefficients` ab, um wie viele Jahre die Lebenserwartung steigt, wenn sich das logarithmierte BIP um 1 erhöht. Das ist die Zahl in der Zeile `log(gdpPercap)`, Spalte `Estimate`.

```r
modell <- lm(lifeExp ~ log(gdpPercap), data = daten_2007)
summary(modell)
```
