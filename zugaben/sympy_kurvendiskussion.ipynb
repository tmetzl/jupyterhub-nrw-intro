# Zugabe: Kurvendiskussion mit SymPy

Bisher haben wir mit Zahlen gerechnet. **SymPy** rechnet mit Symbolen: Es leitet ab, löst Gleichungen und liefert exakte Ergebnisse wie $1 + \sqrt{3}$ statt $2{,}732$. Jupyter zeigt die Formeln dabei als gesetzten Mathe-Satz.

Wir machen eine klassische Kurvendiskussion, wie in der Schule: Ableitungen, Nullstellen, Extrema, Zeichnung.

## Die Funktion

Zuerst sagen wir SymPy, dass `x` ein Symbol ist und keine Zahl. Dann bauen wir daraus eine Funktion. `**` ist in Python das Potenzzeichen.

```python
import sympy as sp

x = sp.symbols("x")

f = x**3 - 3*x**2 + 2
f
```

## Ableitungen

`diff` leitet ab. Das zweite Argument sagt, nach welcher Variablen, das dritte, wie oft.

```python
f1 = sp.diff(f, x)
f2 = sp.diff(f, x, 2)

display(f1, f2)
```

## Nullstellen

`solve` löst die Gleichung $f(x) = 0$. Die Ergebnisse sind exakt.

```python
nullstellen = sp.solve(f, x)
nullstellen
```

Wer Dezimalzahlen braucht, bekommt sie mit `evalf`. Die Zahl in Klammern ist die Anzahl der Stellen.

```python
[n.evalf(4) for n in nullstellen]
```

## Extrema

Kandidaten für Hoch- und Tiefpunkte sind die Nullstellen der ersten Ableitung. Die zweite Ableitung entscheidet: negativ heißt Hochpunkt, positiv heißt Tiefpunkt.

`subs` setzt für `x` einen Wert ein.

```python
kandidaten = sp.solve(f1, x)

for k in kandidaten:
    kruemmung = f2.subs(x, k)
    art = "Hochpunkt" if kruemmung < 0 else "Tiefpunkt"
    print(f"x = {k}: f''(x) = {kruemmung}, also {art} bei ({k}, {f.subs(x, k)})")
```

## Zeichnen

Zum Zeichnen brauchen wir Zahlen statt Symbole. `lambdify` macht aus der Formel eine normale Python-Funktion, die wir an vielen Stellen auswerten und mit matplotlib zeichnen.

```python
import numpy as np
import matplotlib.pyplot as plt

f_num = sp.lambdify(x, f)

xs = np.linspace(-1.5, 3.5, 300)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(xs, f_num(xs), label=f"$f(x) = {sp.latex(f)}$")
ax.axhline(0, color="gray", linewidth=0.8)
ax.scatter([float(n) for n in nullstellen], [0] * len(nullstellen), color="red", zorder=3, label="Nullstellen")
ax.scatter([float(k) for k in kandidaten], [float(f.subs(x, k)) for k in kandidaten], color="green", zorder=3, label="Extrema")
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.legend()
plt.show()
```

**Aufgabe:** Ändere die Funktion ganz oben, zum Beispiel in `x**4 - 4*x**2` oder `x**2 - 2*x - 3`, und führe **Restart Kernel and Run All Cells** aus. Alle Ableitungen, Nullstellen, Extrema und die Zeichnung passen sich an.

Bei Polynomen klappt das immer. Bei anderen Funktionen kann `solve` auch mal keine exakte Lösung finden, dann hilft `sp.nsolve` mit einem Startwert.
