Ein Notebook besteht aus **Zellen**. Es gibt zwei Sorten:

- **Markdown-Zellen** enthalten Text, so wie diese hier.
- **Code-Zellen** enthalten Programmcode, den du ausführen kannst.

Eine Zelle führst du aus, indem du hineinklickst und **Shift + Enter** drückst.
Alternativ geht der Play-Knopf (▶) in der Leiste oben.

Probier es gleich mit der nächsten Zelle aus.

```python
print("Hallo JupyterHub.nrw")
```

Die Ausgabe erscheint direkt unter der Zelle. Das ist das Grundprinzip: Code oben, Ergebnis darunter, beides im selben Dokument.

## Code-Zellen teilen sich ein Gedächtnis

Die nächste Zelle legt eine **Variable** an. Sie gibt nichts aus, das ist Absicht.

```python
anzahl = 100
```

Diese Zelle benutzt die Variable aus der Zelle darüber:

```python
anzahl * 2
```

Die beiden Zellen kennen sich, weil im Hintergrund ein Programm läuft, das sich alles merkt: der **Kernel**. Jede Code-Zelle, die du ausführst, verändert seinen Zustand.

---

## Die Reihenfolge-Falle

Das Gedächtnis des Kernels hat eine Tücke. Probier aus:

**Aufgabe:**

1. Ändere oben `anzahl = 3` in `anzahl = 10`, aber **führe die Zelle nicht aus**.
2. Führe nur die Zelle `anzahl * 2` erneut aus.

Das Ergebnis ist immer noch 6. Der Kernel kennt die Änderung nicht, weil die Zelle nie ausgeführt wurde.

Schau auf die Zahl in eckigen Klammern links neben den Code-Zellen, zum Beispiel `[3]`. Sie zeigt, in welcher **Reihenfolge** die Zellen ausgeführt wurden. Die Reihenfolge im Dokument kann davon abweichen. Genau das macht Notebooks manchmal schwer nachvollziehbar.

---

## Die Lösung: Restart & Run All

Oben im Menü gibt es **Kernel → Restart Kernel and Run All Cells**. Das löscht das Gedächtnis des Kernels und führt alle Zellen von oben nach unten aus.

**Aufgabe:** Führe "Restart Kernel and Run All Cells" aus und schau, was bei `anzahl * 2` jetzt steht.

Wenn ein Notebook nach "Restart & Run All" dasselbe Ergebnis liefert, ist es **reproduzierbar**. Das ist der rote Faden für heute: Wer so ein Notebook weitergibt, gibt das Ergebnis gleich mit.
